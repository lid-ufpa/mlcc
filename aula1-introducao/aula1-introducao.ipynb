## Uma operação fundamental com numpy é a criação de arrays numpy que pode ser feita de diferentes formas: `array()`, `arange()`, `linspace()`, `zeros()`, `ones()` e `full()`.


```python
data1 = np.array([7, 2, 4, 9, 5]) # criar array a partir de uma list de python
data2 = np.arange(0, 5, 1) # números sequenciais com (início, fim não inclusivo, passo)
data3 = np.linspace(0, 5, 4) # números sequências com (início, fim inclusivo, quantidade de números)
data4 = np.zeros(6) # vetor de zeros
data5 = np.ones(5) # vetor de uns
data6 = np.full(7, fill_value = 54) # vetor de um único valor

print(f'Tamanho {data1.shape[0]}. Dados: {data1}')
print(f'Tamanho {data2.shape[0]}. Dados: {data2}')
print(f'Tamanho {data3.shape[0]}. Dados: {data3}')
print(f'Tamanho {data4.shape[0]}. Dados: {data4}')
print(f'Tamanho {data5.shape[0]}. Dados: {data5}')
print(f'Tamanho {data6.shape[0]}. Dados: {data6}')
```

## **Vetorização** corresponde a operações executadas em *batch*. **<font color='red'>Não há necessidade de criar loops</font>**. E as **operações aritméticas** entre arrays de mesmo tamanho são executadas elemento-a-elemento.

```python
# criam-se dois arrays
a = np.array([1, 6, 3])
b = np.array([3, 4, 9])

# operações aritméticas vetorizadas
print(f'Soma de vetores: {a + b}')
print(f'Subtração de vetores: {a - b}')
print(f'Multiplicação de vetores: {a * b}')
print(f'Divisão de vetores: {a / b}')
print(f'Multiplicação por um escalar: {a * 2}')
print(f'Divisão por um escalar: {b / 3}')
print(f'Exponenciação vetorizada: {a ** 4}')
print(f'Filtro vetorizado: {b >= 4}')
```

## Há muitas formas de acessar os elementos de um array e de popular com valores específicos. Por exemplo, o **broadcast** é muito utilizado para atribuir um mesmo valor a várias posições do array. Já o **slice** permite acessar intervalos do array (acesso por referência, **<font color='red'>cuidado!</font>**).

```python
data = np.random.rand(5) # cria um vetor com números aleatórios
print(f'Vetor original: {data}')
data[2:4] = 99 # broadcast
print(f'Vetor após broadcast: {data}')
```

```python
d = data[1:4] # slice (d é uma referência para o vetor original)
d[:] = 77 # broadcast
print(f'Vetor resultante: {data}')
```

## Em arrays bi-dimensionais, os elementos podem ser acessados **recursivamente** ou através de uma lista separada por vírgula, equivalentemente.

```python
m = [[1, 2, 3, 4],
     [5, 6, 7, 8]]  # cria uma lista de listas em python
data = np.array(m)  # criar uma matriz 2x4 em numpy
print(f'Matriz: \n{data}')
print(f'Tamanho: {data.shape}')
```

## Formas de acesso de uma matriz (ou de um array multidimensional) juntamente com mais slices:

```python
print(f'Acesso recursivo de um elemento: {data[1][2]}')
print(f"Acesso separado por 'vírgula': {data[1, 2]}")
print("\n===Mais slices===")
print(data[:2, :1], end = "\n\n")
print(data[:, 3], end = "\n\n")
print(data[-1, :], end = "\n\n")
print(data[:, [1, 3]], end = "\n\n")
print("\n===Copia por máscara===")
r = data[data > 6] # copiou de fato
r[:] = 99 # não surte efeito em data
print(f'Vetor original:\n {data}')
```

## **Line plot**

```python
y = [random.random() for i in range(100)] # pythonic way: criação de uma list em python
plt.figure(figsize=(8, 6))
plt.plot(y, 'o-r') # notação: marcador/linha/cor
plt.show() # mostra a figura na própria célula do jupyter (ou abre uma janela, caso não seja no jupyter)
```

```python
f1 = lambda x: x**2 # função lambda para um polinômio quadrático
x = list(range(10)) # criação de uma list em python para o eixo x
y1 = [f1(i) for i in x] # criação de uma list em python para o eixo y

f2 = lambda x: math.log(x) * 20 # função lambda para uma equação logarítmica
y2 = [f2(i + 1) for i in x] # criação de uma list em python para o eixo y


fig, ax = plt.subplots(figsize = (8, 6))

ax.plot(x, y1, marker = 'o',
               linestyle = '--',
               color = 'k',
               linewidth = 4,
               markersize = 12,
               label = 'f(x) = x^2')


ax.plot(x, y2, marker = '*',
               linestyle = '--',
               color = 'r',
               linewidth = 4,
               markersize = 12,
               label = 'f(x) = log(x) * 20')

fs = 18
ax.set_title('Funções matemáticas', fontsize = fs)
ax.set_xlabel('x', fontsize = fs)
ax.set_ylabel('y', fontsize = fs)
ax.legend(loc = 'upper left', fancybox = True,
           shadow = True, fontsize = fs)

ax.tick_params(axis = 'both', labelsize = fs)

plt.show()
```

## **Bar plot**

```python
y = [150, 150, 100, 300, 320, 280, 70, 80, 60, 150, 180, 200]
x = list(range(len(y)))
labels = ['Arroz', 'Feijao', 'Macarrao', 'Batata', 'Tomate',
          'Cenoura', 'Cafe', 'Leite', 'Acucar', 'Carne',
          'Peixe', 'Frango']

fig, ax = plt.subplots(figsize = (8, 6))

barv = ax.bar(x, y, width = .5,
                    color = 'red',
                    tick_label = labels,
                    edgecolor = 'blue',
                    linewidth = 2)

fs = 16
ax.set_title('Estoque antes das vendas\n', fontsize = fs)
ax.set_xlabel('Produtos', fontsize = fs)

for rect in barv:
    h = rect.get_height()
    ax.text(rect.get_x() + rect.get_width() / 2., h, f'{h}',
            ha = 'center', va = 'bottom', fontsize = fs)

ax.spines[['top', 'left', 'right']].set_visible(False)

ax.tick_params(axis = 'both',
               labelsize = fs,
               left = False,
               labelleft = False,
               labelrotation = 90)
plt.show()
```

```python
y = [150, 150, 100, 300, 320, 280, 70, 80, 60, 150, 180, 200]
x = list(range(len(y)))
labels = ['Arroz', 'Feijao', 'Macarrao', 'Batata', 'Tomate',
          'Cenoura', 'Cafe', 'Leite', 'Acucar', 'Carne',
          'Peixe', 'Frango']

fig, ax = plt.subplots(figsize = (8, 6))

barh = ax.barh(x, list(reversed(y)),
               height = .5,
               color = 'red',
               tick_label = list(reversed(labels)),
               edgecolor = 'blue',
               linewidth = 2)

fs = 16
ax.set_title('Estoque antes das vendas\n', fontsize = fs)
ax.set_ylabel('Produtos', fontsize = fs)

for rect in barh:
    h = rect.get_width()
    t = rect.get_y() + rect.get_height() / 2.
    ax.text(h, t, f'       {h}',
            ha = 'center', va = 'center', fontsize = fs)

ax.spines[['top', 'right', 'bottom']].set_visible(False)

ax.tick_params(axis = 'both',
               labelsize = fs,
               bottom = False,
               labelbottom = False)
plt.show()
```

## **Pizza plot**

```python
vals = [50, 20, 30] # porcentagens: tamanhos das fatias
labels = ['A', 'B', 'C'] # label externo das fatias da pizza
n_labels = [str(i) + ' %' for i in vals] # labels internos da pizza

colors = ['lightcoral','gold', 'lightskyblue'] # nomes das cores
explode = (0.02, 0.3, 0.02) # opção de formatação das fatias (explosão)

wp = {'linewidth' : 1, 'edgecolor' : 'green'} # tamanho das linhas da curva e sua coloração

fig, ax = plt.subplots(figsize = (8, 6))

ax.pie(vals, explode = explode,
             labels = labels,
             colors = colors,
             shadow = True,
             startangle = 150,
             textprops = {'fontsize': 30},
             wedgeprops = wp,
             autopct = lambda x: str(round(x, 2)) + '%')

ax.set_title(label = 'Gráfico de pizza\n',
             fontsize = 22,
             fontweight = 'bold',
             y = 1, # padding superior
             color = '#780050')

plt.show()
```

## ***Series***: array unidimensional contendo uma sequência de valores e um índice contendo rótulos de dados.

## Inicializando Series

```python
dados = pd.Series([4, 7, -5, 3, 10])
print(dados, end = '\n\n')
print(dados.index)
print(dados.values)
```

## Trocando os índices e acessando os valores

```python
dados = pd.Series([4, 7, -5, 3, 10], index = ['b', 'c', 'ola', 'a', 'd'])
print(dados, end = '\n\n')
print(dados['ola'], end = '\n\n')
# print(dados[2], end = '\n\n') # deprecated
print(dados[['ola', 'b', 'd']], end = '\n\n')
```

## Alternativamente, Series também aceita dicionários

```python
d1 = {
    'Bel': 35000,
    'Mos': 15000,
    'Ico': 16000,
    'Out': 10000
}

d2 = {
    'Out': 5000,
    'Ico': 6000,
    'Bel': 5000,
    'Mos': 5000
}
dados1 = pd.Series(d1)
dados2 = pd.Series(d2)
```

## Operações com Series

```python
print(dados1 - dados2, end = "\n\n")
print(dados > 15000, end = "\n\n")
print(dados[dados > 15000], end = "\n\n")
print(dados * 4, end = "\n\n")
print(pow(dados, 2), end = "\n\n")
```

## **Prática de pandas com uma base fictícia de um mercadinho.**

## **<font color='blue'>Carregue os módulos abaixo.</font>**

```python
import pandas as pd
import matplotlib.pyplot as plt
import requests
```

## Baixe a base de dados e visualize-a.

```python
url = "https://raw.githubusercontent.com/regicsf2010/RUUFPa/master/base_de_varejo.csv"
base = pd.read_csv(url)
base
```

## Use a função *info()* para inspecionar os tipos de cada coluna.

```python
base.info()
```

## As células abaixo fazem parte de um trabalho normalmente ministrado para os cursos de computação da UFPa. As células já possuem as respectivas respostas de cada solicitação. Vamos analisar individualmente cada solicitação e a codificação da resposta para fins de estudo.

## 1. Realize um pré-processamento na base de dados para transformar os valores VV e VC em numéricos com separador ‘.’ para casas decimais.

```python
base['VC'] = base['VC'].replace(',', '.', regex = True).astype(float)
base['VV'] = base['VV'].replace(',', '.', regex = True).astype(float)

ms = ['JAN', 'FEV', 'MAR', 'ABR', 'MAI', 'JUN', 'JUL', 'AGO', 'SET', 'OUT', 'NOV', 'DEZ'] # vetor auxiliar
base
```

## 2. Calcule o lucro mensal da empresa pela seguinte equação:
## `lucro_liquido = (lucro_bruto - lucro_liquido) / lucro_bruto`

```python
base['LUCRO'] = (base['VV'] - base['VC']) / base['VV'] # uma nova coluna será criada
```

```python
lucro_mensal = base[ms].apply(lambda x: sum(x * base['VV']  * base['LUCRO']))

fig, ax = plt.subplots(figsize = (12, 6))

ax.plot(ms, lucro_mensal, '-o', label = 'Lucro')

fs = 16
ax.set_title('Lucro geral mensal da empresa', fontsize = fs)
ax.set_xlabel('Mês', fontsize = fs)
ax.set_ylabel('Valor em reais', fontsize = fs)
ax.tick_params(axis = 'both', which = 'major', labelsize = fs)
ax.grid(axis = 'x', color = 'lightblue', linestyle = '--')
ax.legend(loc = 'upper right', bbox_to_anchor = (1.2, 1),
          fancybox = True, shadow = True, fontsize = fs)
plt.tight_layout()
```

## 3. Informe o produto com o maior e o menor lucro para empresa.

```python
idx_max = base['LUCRO'].argmax()
idx_min = base['LUCRO'].argmin()
print(f"Maior lucro: {base.iloc[idx_max]['PROD']} com {base.iloc[idx_max]['LUCRO']:1.3f}% de lucro.")
print(f"Menor lucro: {base.iloc[idx_min]['PROD']} com {base.iloc[idx_min]['LUCRO']:1.3f}% de lucro.")
```

## 4. Plote um gráfico no plano cartesiano contendo nos eixos: (x) meses e (y) quantidade de cada produto vendido.

```python
fig, ax = plt.subplots(figsize = (12, 6))

for i in range(base.shape[0]):
    ax.plot(ms, base.iloc[i][ms], '-o', label = base.iloc[i]['PROD'])

fs = 16
ax.set_title('Número de vendas por mês', fontsize = fs)
ax.set_xlabel('Mês', fontsize = fs)
ax.set_ylabel('Quantidade', fontsize = fs)
ax.tick_params(axis = 'both', which = 'major', labelsize = fs)
ax.grid(axis = 'x', color = 'lightblue', linestyle = '--')
ax.legend(loc = 'upper right', bbox_to_anchor = (1.2, 1),
          fancybox = True, shadow = True, fontsize = fs)
plt.tight_layout()
```

## 5. Plote qual o estoque restante de cada produto após 12 meses.

```python
base['TOTAL_VENDAS'] = base[ms].sum(axis = 1).values
base['ESTOQUE_RESTANTE'] = base['QUANT'] - base['TOTAL_VENDAS']
base.sort_values('ESTOQUE_RESTANTE', ascending = False, inplace = True)
base
```

```python
fig, ax = plt.subplots(figsize = (8, 6))

x_pos = [i * 2 for i in range(base.shape[0])]
barv = ax.bar(x_pos, base['ESTOQUE_RESTANTE'], width = .9, label = 'Produtos')

ax.set_xticks(x_pos, labels = base['PROD'])
ax.tick_params(axis = 'both', which = 'major', labelrotation = 90, labelsize = 'xx-large')
for rect in barv:
    h = rect.get_height()
    ax.text(rect.get_x() + rect.get_width() / 2., 1.01 * h, h, ha = 'center', va = 'bottom', fontsize = 'large')

ax.spines[['top', 'left', 'right']].set_visible(False)
ax.tick_params(left = False, labelleft = False)

plt.tight_layout()
```

## 6. Informe, para cada produto, qual o mês em que houve o maior quantidade de vendas.

```python
for i in range(base.shape[0]):
    print(f"{base.iloc[i]['PROD']}: {ms[base[ms].iloc[i].argmax()]}")
```

## 7. Informe o total investido, vendido e lucro (ou prejuízo) total da empresa após 12 meses.

```python
investido = sum(base['VC'] * base['QUANT'])
print(f'Total investido: R$ {investido}')

vendido = sum(base['VV'] * base['TOTAL_VENDAS'])
print(f'Total vendido: R$ {vendido}')

lucro = vendido - investido
print(f'Total de lucro: R$ {lucro}')
```

## 8. Plote a quantidade de venda de cada produto após 12 meses.

```python
base['TOTAL_VENDAS'] = base[ms].sum(axis = 1).values
base.sort_values('TOTAL_VENDAS', ascending = False, inplace = True)

fig, ax = plt.subplots(figsize = (8, 6))

x_pos = [i * 2 for i in range(base.shape[0])]
barv = ax.bar(x_pos, base['TOTAL_VENDAS'], width = .9, label = 'Produtos')

ax.set_xticks(x_pos, labels = base['PROD'])
ax.tick_params(axis = 'both', which = 'major', labelrotation = 90, labelsize = 'xx-large')
for rect in barv:
    h = rect.get_height()
    ax.text(rect.get_x() + rect.get_width() / 2., 1.01 * h, h, ha = 'center', va = 'bottom', fontsize = 'large')

ax.spines[['top', 'left', 'right']].set_visible(False)
ax.tick_params(left = False, labelleft = False)

plt.tight_layout()
```
