Na base de dados em questão, a coluna "stalk-root" contém 5644 não nulos. Isso significa que há dados faltantes. Nesse processamento, os valores faltantes serão substituídos por "-1".

- Codificar os dados
- Padronizar (normalizar)

````python
# Importar a  biblioteca do scikit-learn
from sklearn.preprocessing import OrdinalEncoder, StandardScaler
from sklearn.model_selection import train_test_split
````

````python
# Criar o codificador com um valor para representar valores ausentes
dados_cogumelos_process = OrdinalEncoder(encoded_missing_value = -1).fit_transform(dados_cogumelos).astype('int')
dados_cogumelos_process
````

Para visualizar como dataframe pode executar o comando:


### 1.5 Normalizar os dados

A normalização de dados consiste em deixar os valores dos atributos em uma escala não discrepante entre eles. Esta etapa é crucial para obter uma estabilidade durante o treinamento do modelo e evitar overflow.

Para essa base de dados usaremos a normalização normal padrão (gaussiana), que deixa os dados com média 0 e desvio padrão 1. Essa normalização é expressa pela equação:

$$z_i=\frac{x_i-\mu}{\sigma}$$

Onde:
* $x_i$ - é o valor do atributo de elemento _i_
* $\mu$ - é a média do atributo
* $\sigma$ - é o desvio padrão


````python
# Separar dados de entrada -> X e saida -> y
X = dados_cogumelos_process[:,:-1]
y = dados_cogumelos_process[:,-1]
````

````python
# Normalizacao standard
entrada = StandardScaler().fit_transform(X)
saida = y.reshape(-1,1)
````

### 1.6 Dividir os dados em treino e teste

Separamos uma parte da base de dados para treinar a nossa RNA e outra parte para testar a perfomance.

Nesse caso usaremos 80% para Treino e 20% para Teste.

````python
# Dividir os dados em conjuntos de treino e teste 80-20
X_treino, X_teste, y_treino, y_teste = train_test_split(entrada, saida, test_size=0.2, random_state=42)
````

As etapas para treinar o modelo são:
- Definir a topologia da RNA (camadas densas, neurônios, funções de ativação, etc.);
- Configurar a RNA (Optimizador + taxa de aprendizagem, função de custo, tamanho do lote de treinamento);
- Treinar o modelo (fit);
- Avaliar (Ver acurácia, perdas, matriz de confusão e outras métricas).


````python
# Importar bibliotecas necessárias
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
````

### 2.1 Definir a Topologia e configurar a rede

````python
# A) Topologia da rede
modelo = Sequential([

      Dense(units = 32, activation = 'relu'),
      Dense(units=64, activation='sigmoid'),
      Dense(units = 1, activation = 'sigmoid')
])
##

# B) Configuracoes da rede
modelo.compile(loss='mse',
                optimizer = Adam(learning_rate=0.001),
                metrics = ['accuracy'])
````

## 2.2 Treinar a RNA

````python
# Treinamento da rede. Alem de passar os dados de entrada X e saida y, definir numero de epocas e tamanho do lote
historico = modelo.fit(X_treino, y_treino, epochs=10, batch_size=32)
````

## 3. Avaliar e testar a RNA

````python
from sklearn.metrics import confusion_matrix, f1_score, accuracy_score, precision_score, recall_score, ConfusionMatrixDisplay
import numpy as np
import matplotlib.pyplot as plt
````

````python
# Avaliar o modelo (mostrar as predicoes para os dados de teste)
def avaliar_modelo(modelo, x_test):
  y_pred = modelo.predict(x_test)
  y_pred = np.where(y_pred>0.5,1,0)

  return y_pred


# Calcular métricas: Acurácia, Precision, Recall, F1 Score
def mostrar_metricas(y_true, y_pred):
    acuracia = accuracy_score(y_true, y_pred)
    f1score = f1_score(y_true, y_pred, average='macro')
    precisao = precision_score(y_true, y_pred, average='macro')
    recall = recall_score(y_true, y_pred, average='macro')

    print(f'Acurácia: {acuracia}')
    print(f'Precision: {precisao}')
    print(f'Recall: {recall}')
    print(f'F1 Score: {f1score}')



# mostrar grafico do erro medio
def mostrar_historico (history):
    plt.plot(history.history['loss'], label='Dados de treino')
    plt.xlabel('Épocas')
    plt.ylabel('Erro médio')
    plt.legend()
    plt.grid()
    plt.show()



# Mostrar matriz de confusao
def mostrar_mc (y_test, y_pred):
  mc = confusion_matrix(y_test, y_pred)
  mc_disp = ConfusionMatrixDisplay(mc)
  mc_disp.plot(cmap='plasma', colorbar=False)
  plt.show()
````

````python
y_pred = avaliar_modelo(modelo, X_teste)
````

````python
# Mostrar as métricas
mostrar_metricas(y_teste, y_pred)
````

````python
mostrar_historico(historico)
````

````python
mostrar_mc(y_teste, y_pred)
````

````python
mostrar_mc(y_teste, y_pred)
````
